```python
#!pip install selenium

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from getpass import getpass
import time

# WebDriverの設定と起動
options = webdriver.ChromeOptions()
options.add_argument('--headless')
options.add_argument('--no-sandbox')
options.add_argument('--disable-dev-s hm-usage')

chrome_driver = webdriver.Chrome(options=options)

# TERAKOYAのトップページにアクセス
chrome_driver.get('https://terakoya.sejuku.net/register')

# ヘッダーのログインボタンをクリック
wait = WebDriverWait(chrome_driver, 30)
header_login_button = wait.until(
    EC.element_to_be_clickable((By.CSS_SELECTOR, '#root > header > div.sc-kNtSYm.cgjwRC'))
)
header_login_button.click()

# メールアドレスとパスワードを入力
email_address = input('メールアドレスを入力してください：')
password = getpass('パスワードを入力してください: ')

# メールアドレスとパスワードの入力欄を見つける
parent_element = wait.until(
    EC.presence_of_element_located((By.XPATH, '//*[@id="root"]/div[1]/div[1]/div[2]/div'))
)
email_input = parent_element.find_element(By.NAME, 'email')
password_input = parent_element.find_element(By.NAME, 'password')

email_input.send_keys(email_address)
password_input.send_keys(password)

# フォームのログインボタンをクリック
form_login_button = wait.until(
    EC.element_to_be_clickable((By.XPATH, '//*[@id="root"]/div[1]/div[1]/div[2]/button'))
)
form_login_button.click()

# ログイン成功後の要素が表示されるまで待つ
wait.until(EC.visibility_of_element_located((By.CSS_SELECTOR, 'nav > img')))


print("ログイン成功！")

# ログイン成功後、アカウント設定ページに遷移
chrome_driver.get('https://terakoya.sejuku.net/account/profile')

time.sleep(5)


chrome_driver.save_screenshot('screenshot.png')

# 編集ボタンをクリック
edit_button = wait.until(
    EC.element_to_be_clickable((By.XPATH, '//*[@id="root"]/div[1]/div/div/main/div/div[1]/button'))
)
edit_button.click()

# 自己紹介のテキストエリアを見つけて入力
introduction_textarea = wait.until(
    EC.presence_of_element_located((By.XPATH, '//*[@id="root"]/div[1]/div/div/main/div/div[1]/div[9]/div[2]/textarea'))
)
introduction_textarea.clear()  # 既存のテキストをクリア
introduction_textarea.send_keys('プログラミング学習中です！今はスクレイピングに挑戦しています！')

# 画面キャプチャを保存
chrome_driver.save_screenshot('profile_edit.png')

# 更新ボタンをクリック
update_button = wait.until(
    EC.element_to_be_clickable((By.XPATH, '//*[@id="root"]/div[1]/div/div/main/div/div[1]/button[2]'))
)
update_button.click()

print("自己紹介を更新しました！")

# ブラウザを閉じる
chrome_driver.quit()
```
